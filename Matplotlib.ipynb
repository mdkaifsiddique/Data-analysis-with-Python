```python
import numpy as np
```

```python
#Part 1
```

```python
x=np.linspace(0,5,15)
y=x**2
```

```python
x
```

```python
y
```

```python
#Functional
```

```python
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('line chart')
```

```python
plt.plot(x,y,'g')   # 'g' for color
```

```python
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title('Blue line chart')

plt.subplot(1,2,2)
plt.plot(x,y,'r')
plt.title('Red line chart')
```

```python
#Object Oriented
```

```python
fig=plt.figure()
axes=fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y,'r')

axes.set_xlabel('x-axis')
axes.set_ylabel('y-axis')
axes.set_title('Line chart')
```

```python
fig=plt.figure()

axes1=fig.add_axes([0.1,0.1,0.8,0.8])
axes2=fig.add_axes([0.4,0.3,0.5,0.2])

axes1.plot(x,y)
axes2.plot(y,x,'r')
```

```python
#Part-2(Sub plot)
```

```python
fig,axes = plt.subplots(nrows=2, ncols=1)

axes[0].plot(x,y)
axes[0].set_title('First chart')
axes[0].set_xlabel('X-axis')
axes[0].set_ylabel('y-axis')

axes[1].plot(y,x,'r')
axes[1].set_title('Second chart')
axes[1].set_xlabel('X-axis')
axes[1].set_ylabel('y-axis')

#plt.tight_layout
```

```python
#Figure size & DPI
```

```python
fig=plt.figure(figsize=(8,3),dpi=100)

ax=fig.add_axes([0,0,1,1])
ax.plot(x,y)

ax.set_xlabel('x-axis')
ax.set_ylabel('y-axis')
```
