#### Machine Learning

when we have target value in category then we use logistic
when we have target value in numeric then we use linear reg

```python
# supervised learning
from sklearn.linear_model import LinearRegression
```

```python
iris=load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
y
```

```python
model=LinearRegression()
```

```python
model.fit(X,y)
```

```python
#if upper command was not run then we use below command

model.intercept_
```

```python
model.coef_
```

```python
#Model evaluation

y_pred=model.predict(X)
```

```python
y_pred
```

```python
r_squared = model.score(X,y)
r_squared
```
