```python
import numpy as np
```

### Part 1

```python
x=np.linspace(0,5,15)
y=x**2
```

```python
x
```

```python
y
```

```python
#Functional
```

```python
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('line chart')
```

```python
plt.plot(x,y,'g')   # 'g' for color
```

```python
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title('Blue line chart')

plt.subplot(1,2,2)
plt.plot(x,y,'r')
plt.title('Red line chart')
```

```python
#Object Oriented
```

```python
fig=plt.figure()
axes=fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y,'r')

axes.set_xlabel('x-axis')
axes.set_ylabel('y-axis')
axes.set_title('Line chart')
```

```python
fig=plt.figure()

axes1=fig.add_axes([0.1,0.1,0.8,0.8])
axes2=fig.add_axes([0.4,0.3,0.5,0.2])

axes1.plot(x,y)
axes2.plot(y,x,'r')
```

### Part-2(Sub plot)

```python
fig,axes = plt.subplots(nrows=1, ncols=2)

axes[0].plot(x,y)
axes[0].set_title('First chart')
axes[0].set_xlabel('X-axis')
axes[0].set_ylabel('y-axis')

axes[1].plot(y,x,'r')
axes[1].set_title('Second chart')
axes[1].set_xlabel('X-axis')
axes[1].set_ylabel('y-axis')

#plt.tight_layout
```

```python
#Figure size & DPI
```

```python
fig=plt.figure(figsize=(8,3),dpi=100)

ax=fig.add_axes([0,0,1,1])
ax.plot(x,y)

ax.set_xlabel('x-axis')
ax.set_ylabel('y-axis')
```

```python
fig,axes=plt.subplots(nrows=2,ncols=1,figsize=(7,3))

axes[0].plot(x,y)
axes[1].plot(y,x)
```

```python
#legend

fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,x**2,label='x-squred')
ax.plot(x,x**3,label='x-cube')

ax.legend()       #by defult .legend(loc=0); loc=0 best position,, loc vary  0 to 10
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,x**2,label='x-squred')
ax.plot(x,x**3,label='x-cube')

ax.legend(loc=1)
```

```python
#save figure
fig
fig.savefig('line graph.jpg',dpi=200)   #u can save .jeg, .png etc
```

### Part-3

```python
#PLot Apperance
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple")
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple",linewidth=3)     #Here linewidth=lw; we can write any one linewidth or lw, Bydefault lw=1
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple",lw=2,alpha=0.4)     #Here alpha is color apperance & alpha lies between 0 to 1.
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple",lw=2,linestyle='--')   #linestyle = ls 
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple",lw=2,ls=':')    #Bydefault ls="-"
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple",lw=1, ls='-',marker="o")  # we can use marker = "+" , "*" , "o" etc
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])

ax.plot(x,y,color="brown",
        lw=3,
        ls="-",
        marker="*",
        markersize="15",
        markerfacecolor="purple",
        markeredgewidth="2",
        markeredgecolor="green")
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple",
        lw="2",
        ls='--')

ax.set_xlim([0,1])
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,y,color="purple",
        lw="2",
        ls='--')

ax.set_ylim([0,2])
```
