**Matplotlib is the most popular Python library for creating graphs, charts, and visualizations. It is widely used in data analysis, data science, and machine learning**

### Key

 - Matplotlib is excellent 2D and 3D graphical library for scientific figures.
 - Matplotlib gives you control over every aspect of a figure.
 - It was designed to have a similar feel to MATLAB's graphical plotting.
 - It supports many chart types \& Highly customizable.
 - It Integrates with NumPy and Pandas.
 - It produces high-quality visualizations.

### Importing:

import the matplotlib.pyplot module under the name plt(the tidy way)

```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline    #for jupyter notebook

#plt.show()           #for other editor
```

```python
import numpy as np
```

```python
x=np.linspace(0,5,15)
y=x**2
```

```python
x
```

```python
y
```

```python
#Functional
```

```python
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('line chart')
```

```python
plt.plot(x,y,'g')   # 'g' for color
```

```python
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title('Blue line chart')

plt.subplot(1,2,2)
plt.plot(x,y,'r')
plt.title('Red line chart')
```
