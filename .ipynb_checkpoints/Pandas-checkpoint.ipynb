# • Series

A Series is a one-dimensional labeled array in Pandas.\
**A Series contains:** 
 - **Values** — the actual data
 - **Index** — labels used to identify each value

```python
import pandas as pd
import numpy as np
```

```python
# creating a series

dat=[10,20,30,40,50]
pd.Series(dat)
```

```python
labels=["a","b","c","d","e"]
```

```python
print(pd.Series(data=dat,index=labels))
```

```python
#or
print(pd.Series(dat,labels)) #both codes are same
```

```python
#Creating a series from a dictionary

dic={'math':74,'statistics':75,'python':80,'R':90}
ser=pd.Series(dic)
print(ser)
```

**⇑ The dictionary keys become the index and the dictionary values become the data.**

```python
#In Pandas Series,you can extract values using both index/labels and positions,but the method depends on what you mean.

#Simple rule:
#.loc[] → use index labels
#.iloc[] → use integer positions
```

```python
#Accessing Series Values
print(ser["math"])
```

# • Missing Values

**In pandas, missing values represent data that is unavailable or unknown. They are commonly represented as:**

 - NaN (Not a Number) for numeric data
 - None for object/string data

```python
import numpy as np
import pandas as pd
```

```python
d={'A':[1,2,np.nan],'B':[5,np.nan,np.nan],'C':[1,2,3]}
```

```python
df=pd.DataFrame(d)
df
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df.dropna
```

```python
df.dropna(thresh=2)   #returen two rows
```

```python
df
```

```python
df.fillna(value='FILL VALUE')
```

```python
df['A'].fillna(value=df['A'].mean())   #extract col A and fill na value by its mean.
```

```python
df.fillna(value=df['C'].mean())    #fill all NA value by mean of col 'C"
```

# • GroupBy

**Groupby allows you to group together rows based off of a column and perform an aggregate function on them.**

**The groupby() method is used to split data into groups based on one or more columns, perform calculations on each group, and combine the results. It is one of the most powerful features of pandas for data analysis.**

```python
data={'Company':['GOOGLE','GOOGLE','MICROSOFT','MICROSOFT','FACEBOOK','FACEBOOK'],
      'Persion':['Sam','Charlie','Amy','Ansh','Carl','Sarah'],
      'Sales':[200,120,340,124,243,350]}
```

```python
df=pd.DataFrame(data)
df
```

```python
df.groupby('Company')
```

```python
byCompany=df.groupby('Company')
```

```python
byCompany.sum()
```

```python
df
```

```python
byCompany=df.groupby('Company')
byCompany['Sales'].mean()
```

```python
byCompany['Sales'].std()
```

```python
byCompany.sum().loc['FACEBOOK']
```

```python
byCompany.count()
```

```python
byCompany.max()
```

```python
byCompany.min()
```

```python
byCompany.describe()
```

```python
byCompany.describe().transpose()
```

```python
byCompany.describe().transpose()['FACEBOOK']
```

```python
#or
df.groupby('Company').describe().transpose()['GOOGLE']
#byCompany.describe().transpose()['GOOGLE']
```
