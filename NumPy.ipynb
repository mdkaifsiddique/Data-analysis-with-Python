## NumPy
 - **Array:** An array is a data type structure used to store multiple values of the same data type in single variable.

### NumPy
 - used for numerical computing.
 - It provides support for large, multi-dimensional arrays and matrices, along with a wide range of mathematical functions to operate on them efficiently.
 - Faster than Python lists for numerical operations.
 - Uses less memory.
 - (performing operations on entire arrays without loops.
 - Provides many mathematical, statistical, and linear algebra functions.

# NumPy Indexing \& Selection

### Indexing: means accessing one or more elements from a NumPy array using their position (index).
### Selection: means choosing multiple elements using slicing, conditions, or fancy indexing.

```python
import numpy as np
```

```python
arr=np.arange(0,11)
```

```python
print(arr)
```

```python
print(arr[8])
```

```python
print(arr[1:5])
```

```python
print(arr[:6])

#or

# arr[0:6]
```

```python
arr[5:]
```

```python
print(arr[-1])
```

```python
arr[0:5]=100
print(arr)
```

```python
print(arr[0:8:2])     #arr[start:stop:step]
```

```python
#Slicing


ar=np.arange(0,16)
print(ar)
```

```python
slice_of_ar=ar[0:8]
print(slice_of_ar)
```

```python
slice_of_ar[:]=75
```

```python
slice_of_ar
```

```python
ar[::2]    #every second element
```

```python
ar_copy=ar.copy()
```

```python
ar_copy
```

```python
ar_copy[:]=100
```
