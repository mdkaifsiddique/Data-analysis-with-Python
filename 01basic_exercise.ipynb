**Q8 Use the lamda expression and the filter function() to filter out words from a list that don't start with letter 's'. For example**\
['soup','dog','salad','cat','great']\
**should be filtered down**\
['soup','salad']

```python
lst=['soup','salad','dog','cat','great']

print(list(filter(lambda x: x.startswith("s"),lst)))
```

**Q9 You  are driving little too fast, and a police officer stops you.Write a function return one of three possible relusts: "no ticket","small ticket" or "big ticket.If your speed is 60 or less then the relust is "no ticket",if speed between 61 and 80 inclusive, the result is "small ticket",if speed 81 or more then the relust is "big ticket".Unless it is your birthday(encoded as a boolean value in the parameter of teh function) --on your birthday, your speed can be 5 higher in all cases.**

```python
def coughtspeeding(speed,birthday):
    if birthday:
        speed=speed-5

    if speed<=60:
        return "no ticket"
    elif speed<=80:
        return "small ticket"
    else:
        return "big ticket"
```

```python
coughtspeeding(65,False)
```

```python
coughtspeeding(95,True)
```
